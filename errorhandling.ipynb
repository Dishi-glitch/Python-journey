```python
revenue=0
expenses=10

try:
    margin= (revenue- expenses)* 100/ revenue
    print(f"Margin: {margin: .2f}%")

#Division by zero is not possible
except ZeroDivisionError as e : #storing in an obj
    print("Revenue=0 so skipping")    
```
