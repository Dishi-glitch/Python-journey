```python
financials = {
    'Q1': {"revenue": 50, "expenses": 10},
    'Q2': {"revenue": -60, "expenses": 90}
}

for quarter, data in financials.items():
    if data['revenue'] <= 0:  # Also handles division by zero if revenue is 0
        print('skipping', quarter)
        continue
    
    margin = (data['revenue'] - data['expenses']) / data['revenue'] * 100
    print(f"{quarter}: {margin:.2f}%")
```

```python
# FUNCTIONS
                  #type hint to hint abt the data type
def calculate_m_p(revenue: int, expenses: int) -> int 
    profit= revenue - expenses
    margin = profit*100/ revenue
    
    return margin, profit
    
```

```python
def total_expenses(rent, phone_bill):
    return rent+ phone_bill
total_expenses(1000, 30)
```

```python
def total_expenses(**kargs):
    total = 0
    for key, value in kargs.items():
        total += value
    return total

total_expenses(rent=1000, phone_bill=30)    
```

```python
class employee:

    def __init__(self, name, salary, number):
        self.name= name
        self.salary= salary
        self.num= number

    def calculate_bonus(self):
        return self.salary* 0.1

emp1= employee("Dishi", 300000, 7246362329)
emp1.num        
        
```

```python
from datetime import date

class human:
    def __init__(self, name, birth_yr):
        self.name= name
        self.year= birth_yr

    def get_age(self):
            current_year= date.today().year
            return current_year-self.year
    def introduce(self):
         print(f"Hi I am {self.name}, I am {self.get_age()} and l live in London.")

Dishi = human("Dishi", 2008)
Dishi.get_age()
Dishi.introduce()         
```
